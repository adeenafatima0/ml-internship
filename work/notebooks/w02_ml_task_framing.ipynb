## 5. Why ML beats a fixed rule here

From last week's numbers, over half the pages (54.2%) are declining, but my simple hand-rule (stale + still visible) only caught 0.1% of pages as candidates. That's a huge gap. A hand-written rule can only really check a couple things at once with fixed cutoffs, like "180+ days old AND 500+ impressions" — it can't really weigh a bunch of weaker signals together the way a model can. Decline is probably caused by some mix of age, position, CTR, word count, etc. all interacting, and that's just too messy to hand-write as clean if/else logic. This lines up with what I saw in the earlier notebooks too — the random forest got roughly 3x the precision of the hand-rule, so a model clearly picks up on stuff a simple rule misses.
