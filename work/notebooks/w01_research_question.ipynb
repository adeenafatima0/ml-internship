## 3. Quick look at the data (2-3 real numbers)
Across the starter dataset, 54.2% of pages are currently in decline — a substantial pool that could benefit from review. However, a simple hand-written rule (stale AND still visible) only flags 0.1% of pages as candidates, showing that a rigid one-size-fits-all rule is far too narrow to catch most real opportunities. With an average of 5,200 impressions per page, most pages do have meaningful visibility, meaning review effort spent here is unlikely to be wasted on invisible content. This gap — a large declining population, but a tiny rule-based candidate pool — is exactly why a learned ranking model is worth building: to surface a wider, better-prioritized set of real candidates than a fixed rule ever could.

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/adeenafatima0/ml-internship"
REPO_DIR = "ml-internship"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)

print("Working dir:", os.getcwd())
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining_rate = (df["trend_direction"].str.lower() == "down").mean()
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).mean()
avg_impressions = df["impressions_90d"].mean()

print(f"Share of pages currently declining: {declining_rate:.1%}")
print(f"Share of pages that are stale AND still visible (review candidates): {stale_visible:.1%}")
print(f"Average impressions per page (90d): {avg_impressions:.0f}")
```
